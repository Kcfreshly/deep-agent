```python
from pathlib import Path
import sys
from dotenv import load_dotenv

load_dotenv()

project_root = Path.cwd().resolve()
if not (project_root / "advisor.py").exists():
    project_root = project_root / "my-agent"

expected_python = project_root / ".venv" / "Scripts" / "python.exe"
if Path(sys.executable).resolve() != expected_python.resolve():
    raise RuntimeError(
        "Select the 'Python 3 (my-deep-agent)' kernel, restart it, and run again."
    )
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from session import PersistentAdvisorSession

session = PersistentAdvisorSession()
```

```python
def ask_advisor(prompt: str) -> str:
    """Send and persist one conversational turn."""
    return session.ask(prompt)
```
